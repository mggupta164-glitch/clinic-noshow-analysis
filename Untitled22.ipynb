```python
clinics_df = pd.read_excel('/content/Clinic.xlsx')

def fetch_clinic_temp(clinic_name, lat, lon):
    url = "https://open-meteo.com"
    params = {
        "latitude": lat,
        "longitude": lon,
        "hourly": "temperature_2m",
        "forecast_days": 1
    }

    last_reason = "Unknown error"

    # Attempt up to 2 times (initial attempt + 1 retry)
    for attempt in range(2):
        try:
            response = requests.get(url, params=params, timeout=10)

            if response.status_code == 200:
                data = response.json()

                # Check if 'hourly' and 'temperature_2m' exist and are non-empty
                if "hourly" in data and "temperature_2m" in data["hourly"] and len(data["hourly"]["temperature_2m"]) > 0:
                    return data["hourly"]["temperature_2m"][0]
                else:
                    last_reason = "Missing 'hourly' or 'temperature_2m' data in JSON response"
            else:
                last_reason = f"HTTP status code {response.status_code}"

        except requests.exceptions.RequestException as e:
            last_reason = f"RequestException ({e})"

    # If both attempts fail, print the clinic's name and reason, and return None
    print(f"Failed for clinic '{clinic_name}': {last_reason}")
    return None
```

```python
def format_row(row):
    items = []
    for value in row:
        if pd.isna(value):
            items.append("NULL")             # Empty value -> NULL
        elif isinstance(value, (int, float)):
            items.append(str(value))         # Number -> keep without quotes
        else:
            items.append(f"'{value}'")       # Text -> wrap in single quotes
    return ",".join(items)

# Open load_data.sql and write the INSERT lines
with open("load_data.sql", "w") as file:

    # 1. Write clinics rows first
    for _, row in clinics_df.iterrows():
        file.write(f"INSERT INTO clinics VALUES ({format_row(row)});\n")

    # 2. Write appointments rows second
    for _, row in df_clean.iterrows():
        file.write(f"INSERT INTO appointments VALUES ({format_row(row)});\n")

print("Done! load_data.sql is ready.")
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
)
from sklearn.model_selection import (
    GridSearchCV,
    cross_val_predict,
    cross_val_score,
)
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier

# ==========================================
# (a) GridSearchCV for Random Forest on X_train
# ==========================================
param_grid = {
    "n_estimators": list((100, 200)),
    "max_depth": list((3, 5, 8)),
}
grid_rf = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring="roc_auc",
)
grid_rf.fit(X_train, y_train)

print("=== (a) Tuned Random Forest (GridSearchCV) ===")
print("Best Params:", grid_rf.best_params_)
print(f"Best CV ROC-AUC: {grid_rf.best_score_:.4f}")

# ==========================================
# (b) 5-Fold CV ROC-AUC Comparison (Training Set Only)
# ==========================================
lr_pipe = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
dt_depth3 = DecisionTreeClassifier(max_depth=3, random_state=42)
rf_default = RandomForestClassifier(n_estimators=100, random_state=42)

candidates = {
    "Logistic Regression (Pipeline)": lr_pipe,
    "Decision Tree (max_depth=3)": dt_depth3,
    "Random Forest (Default)": rf_default,
    "Random Forest (Tuned)": grid_rf.best_estimator_,
}

print("\n=== (b) 5-Fold Cross-Validated ROC-AUC (X_train) ===")
cv_scores = {}
for name, clf in candidates.items():
  score = cross_val_score(clf, X_train, y_train, cv=5, scoring="roc_auc").mean()
  cv_scores[name] = score
  print(f"  {name}: {score:.4f}")

winner_name = max(cv_scores, key=cv_scores.get)
winner_model = candidates[winner_name]
print(
    f"\n--> Winning Model: {winner_name} (CV ROC-AUC ="
    f" {cv_scores[winner_name]:.4f})"
)

# ==========================================
# (c) Choose Threshold via Out-of-Fold Probabilities
# ==========================================
oof_probs = cross_val_predict(
    winner_model, X_train, y_train, cv=5, method="predict_proba"
)[:, 1]

print(f"\n=== (c) Out-of-Fold Threshold Comparison ({winner_name}) ===")
f1_results = {}
for thresh in (0.5, 0.35):
  preds = (oof_probs >= thresh).astype(int)
  prec = precision_score(y_train, preds)
  rec = recall_score(y_train, preds)
  f1 = f1_score(y_train, preds)
  flagged = preds.sum()
  f1_results[thresh] = f1
  print(
      f"Threshold {thresh:.2f} -> Precision: {prec:.4f} | Recall: {rec:.4f} |"
      f" F1: {f1:.4f} | Flagged: {flagged}"
  )

best_thresh = max(f1_results, key=f1_results.get)
print(
    f"\n--> Chosen Threshold: {best_thresh:.2f} (Higher F1 ="
    f" {f1_results[best_thresh]:.4f})"
)

# ==========================================
# (d) Fit Winner on X_train & Evaluate ONCE on X_test
# ==========================================
winner_model.fit(X_train, y_train)

test_probs = winner_model.predict_proba(X_test)[:, 1]
test_preds = (test_probs >= best_thresh).astype(int)

print(
    f"\n=== (d) Final Test Set Evaluation ({winner_name} @ threshold"
    f" {best_thresh:.2f}) ==="
)
print(f"Test Precision:      {precision_score(y_test, test_preds):.4f}")
print(f"Test Recall:         {recall_score(y_test, test_preds):.4f}")
print(f"Test Number Flagged: {test_preds.sum()} / {len(y_test)}")
print(f"Test ROC-AUC:        {roc_auc_score(y_test, test_probs):.4f}")

# ==========================================
# (e) Save test_risk_scores.csv
# ==========================================
output_df = pd.DataFrame({
    "appointment_id": df.loc[X_test.index, "appointment_id"],
    "no_show": y_test.values,
    "risk_prob": test_probs,
    "flag": test_preds,
})

output_df.to_csv("test_risk_scores.csv", index=False)
print("\n=== (e) Saved test_risk_scores.csv ===")
print(output_df.head())
```

```python
import time
import json
import pandas as pd
from google.genai import types

# List of available Flash models from your account (ordered from lightest/fastest)
CANDIDATE_MODELS = [
    "gemini-3.1-flash-lite",
    "gemini-3.6-flash",
    "gemini-3.5-flash",
    "gemini-3.5-flash-lite",
    "gemini-flash-latest",
    "gemini-3.8-flash",
]

SCHEMA = {
    "type": "object",
    "properties": {
        "root_cause": {"type": "string"},
        "confidence": {"type": "number"},
        "summary": {"type": "string"},
        "recommended_action": {"type": "string"},
    },
    "required": ["root_cause", "confidence", "summary", "recommended_action"],
}

SYSTEM_TEXT = (
    "You are an experienced healthcare clinic operations analyst. "
    "Classify patient appointment no-show notes accurately into the specified JSON structure."
)

def analyse_note(note_text):
    prompt = f"""
Context: A medical clinic collects brief staff follow-up notes explaining why patients missed scheduled appointments.
Role: Act as a healthcare clinic operations analyst reviewing patient no-show records.
Action: Read the note below and extract:
1. root_cause: Must be strictly one of these exact strings: forgot_or_no_reminder, distance_or_transport, cost_or_fee, illness_or_emergency, work_or_schedule, other.
2. confidence: A number between 0 and 1 indicating how confident you are in the root_cause classification.
3. summary: A concise summary of the note in at most 25 words.
4. recommended_action: A practical next step for the clinic front desk to reduce future no-shows for this patient.
Tone: Objective, concise, and operational.
Format: Return valid JSON matching the required schema.

Patient Note:
{note_text}
"""
    for model_name in CANDIDATE_MODELS:
        try:
            response = client.models.generate_content(
                model=model_name,
                contents=prompt,
                config=types.GenerateContentConfig(
                    system_instruction=SYSTEM_TEXT,
                    temperature=0,
                    max_output_tokens=1000,
                    response_mime_type="application/json",
                    response_schema=SCHEMA,
                ),
            )
            print(f"Success using model: {model_name}")
            return json.loads(response.text)
        except Exception as e:
            print(f"{model_name} unavailable ({type(e).__name__}), trying next model...")
            time.sleep(1)
    raise RuntimeError("All candidate models are temporarily overloaded. Please wait 30 seconds and re-run.")

# Load Notes.csv and extract note N001 as a plain string
notes_df = pd.read_csv("/content/Notes.csv")
note_n001 = notes_df.loc[notes_df["note_id"] == "N001", "note_text"].values

result_n001 = analyse_note(note_n001)
print("\nParsed Dictionary for N001:")
print(json.dumps(result_n001, indent=2))
```

```python
import time
import json
import pandas as pd
from pydantic import BaseModel, Field, ValidationError

# (a) Define Pydantic model and allowed root_cause list
ALLOWED_ROOT_CAUSES = {
    "forgot_or_no_reminder",
    "distance_or_transport",
    "cost_or_fee",
    "illness_or_emergency",
    "work_or_schedule",
    "other",
}

class FollowupInsight(BaseModel):
    root_cause: str = Field(min_length=3)
    confidence: float = Field(ge=0, le=1)
    summary: str = Field(min_length=5)
    recommended_action: str = Field(min_length=5)

# (b) Read the notes CSV and loop over the first 10 rows
notes_df = pd.read_csv("/content/Notes.csv").head(10)
results = []

for idx, row in notes_df.iterrows():
    note_id = row["note_id"]
    appointment_id = row["appointment_id"]
    note_text = row["note_text"]

    # Input validation: check pd.isna() first, then empty/whitespace string
    if pd.isna(note_text) or str(note_text).strip() == "":
        results.append({
            "note_id": note_id,
            "appointment_id": appointment_id,
            "root_cause": "other",
            "confidence": 0.0,
            "summary": "No note recorded",
            "recommended_action": "Re-call patient",
            "validation_status": "skipped_blank",
            "raw_response": "",
        })
        print(f"[{note_id}] Skipped blank note.")
        continue

    # Call analyse_note, validate with Pydantic + allowed list, never repair invalid replies
    raw_reply = ""
    data = {}
    validation_status = "invalid"

    try:
        data = analyse_note(str(note_text))
        raw_reply = json.dumps(data)

        # Validate with Pydantic model and allowed-list check
        validated = FollowupInsight(**data)
        if validated.root_cause not in ALLOWED_ROOT_CAUSES:
            raise ValueError(f"Invalid root_cause: {validated.root_cause}")

        validation_status = "valid"
    except (ValidationError, ValueError, TypeError, json.JSONDecodeError, Exception) as e:
        validation_status = "invalid"
        if not raw_reply and isinstance(data, dict) and data:
            raw_reply = json.dumps(data)
        elif not raw_reply:
            raw_reply = str(e)

    # Copy whichever fields were returned; leave missing fields empty ("")
    results.append({
        "note_id": note_id,
        "appointment_id": appointment_id,
        "root_cause": data.get("root_cause", "") if isinstance(data, dict) else "",
        "confidence": data.get("confidence", "") if isinstance(data, dict) else "",
        "summary": data.get("summary", "") if isinstance(data, dict) else "",
        "recommended_action": data.get("recommended_action", "") if isinstance(data, dict) else "",
        "validation_status": validation_status,
        "raw_response": raw_reply,
    })
    print(f"[{note_id}] Processed -> status: {validation_status}")

    # (c) Pause 13 seconds between API calls to respect free-tier rate limit
    if idx < len(notes_df) - 1:
        time.sleep(13)

# (d) Save to notes_analysed.csv with the exact required columns
output_cols = [
    "note_id",
    "appointment_id",
    "root_cause",
    "confidence",
    "summary",
    "recommended_action",
    "validation_status",
    "raw_response",
]
analysed_df = pd.DataFrame(results)[output_cols]
analysed_df.to_csv("/content/notes_analysed.csv", index=False)

print("\nSaved notes_analysed.csv! Summary of validation_status:")
print(analysed_df["validation_status"].value_counts())
display(analysed_df)
```
